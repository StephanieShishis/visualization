What software did you use to create your data visualization?

- I used excel to create my data visualization. For the excel visualization I created a horizontal bar graph by using pivot tables and summarizing the yearly totals.  

Who is your intended audience? 

- My intended audience is the general public or citizens of Toronto who are familiar or know about the Toronto Island Ferry. They would have no prior technicial or background knowledge about the Ferry systems themselves. 
    
What information or message are you trying to convey with your visualization? 

- I want to convey how ferry ridership (through ticket sales and redemption) changed year to year and how a major event like COVID-19 in 2020 impacted the pattern.
    
What aspects of design did you consider when making your visualization? How did you apply them? With what elements of your plots? 

- For the bar graph (like the line graph) I wanted to make it simple and really clear to read. I used the same two distinct and clear colours that are colour-blind friendly (blue and orange) as well as having a clear legend, clear axis titles and numbers. I also used fonts and typefaces that are easy to read. In addition, I inlcuded gridlines for easy value comparison. I also chose a horizontal bra graph so all the years could be read top to bottm without crowding the year labels. Although the excel visualization is clear and understandable, I do think the python visualization is easier to visualize and see trends, espeically the dip in 2020 with COVID-19.

How did you ensure that your data visualizations are reproducible? If the tool you used to make your data visualization is not reproducible, how will this impact your data visualization? 

- The excel visualization is not as reproducible as the python version as it requires some reformatting of the Timestamp column using excel formulas (DATEVALUE, LEFT) to create a new year column that just displays the year. In addition, building of a pivot table and then a pivot chart is needed, which cannot be saved as a repeatable script. Without this understanding, it can be quite difficult to reproduce the graphs as a prioir understanding of how to make pivot charts and graphs is needed. But, once this knowledge is gained, it is easy to use the excel file and plug/play the reuqired data points into the chart. Changing the design formatting (colours, fonts, sizes etc.) can also be an issue as Excel defaults to its standard style unless manually changed. 
    
How did you ensure that your data visualization is accessible?  
    
- I used colourblind-friendly colours (blue and orange) that were also bright and I labelled axes clearly with readbale font sizes. I could improve the visualization further by adding alt-text or have specific data labels on each bar. 

Who are the individuals and communities who might be impacted by your visualization?  

- Some individuals and communities who might be impacted by my visualization include ferry commuters and tourists for service freqeuncy and availbility, city planners for infrastructure and planning and busnisses on the island which could impact sales and foot traffic. 
    
How did you choose which features of your chosen dataset to include or exclude from your visualization? 

- I included important metrics to understand the ridership which was timestamp (coverted to year), sales count and redemptions count. The id column was just a row number and had no meaning so I excluded this. 
    
What ‘underwater labour’ contributed to your final data visualization product?

- Some of the aspects that were hidden work was realizing that the Timestamp column in the CSV was stored as a text value instead of a date value so it had be reformatted. Also, figuring out the DATEVALUE/LEFT formula to seperate the year was a little difficult and troubleshooting what Excel could "group" was a little bit of a hurdle. Lastly, working through the pivot table functions was also a learning curve that took some time.
