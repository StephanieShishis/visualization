What software did you use to create your data visualization?

- I used python to create my data visualization. 

Who is your intended audience? 

- General public or citizens of Toronto who are familiar or know about the Toronto Island Ferry. Have no prior technicial or background knowledge. 
    
What information or message are you trying to convey with your visualization? 

- I want to convey how ferry ridership (through ticket sales and redemption) changed year to year and how a major event like COVID-19 in 2020 impacted the pattern.
    
What aspects of design did you consider when making your visualization? How did you apply them? With what elements of your plots? 

- For both visualizations I wanted to make it simple and really clear to read. I used two distinct and clear colours that are colour-blind friendly (blue and orange) as well as having a clear legned, clear axis titles and numbers. I also used fonts and typefaces that are easy to read. 

How did you ensure that your data visualizations are reproducible? If the tool you used to make your data visualization is not reproducible, how will this impact your data visualization? 

- The python script is fully reproducible. Anyone who has the same CSV file and python script I created can run it and make the same chart. I also documented steps with comments in the code to ensure this reproducibility. 
    
How did you ensure that your data visualization is accessible?  
    
- I used colourblind-friendly colours (blue and orange) that were also bright, I labelled axes clearly with readbale font sizes and I could improve further by adding alt-text. 

Who are the individuals and communities who might be impacted by your visualization?  

- Some individuals and communities include ferry commuters and tourists for service freqeuncy and availbility, city planners for infrastructure and planning and busnisses on the island which could impact sales and foot traffic. 
    
How did you choose which features of your chosen dataset to include or exclude from your visualization? 

- Included important metrics to understand the ridership which was timestamp (year), sales count and redemptions count. The id column was just a row number and had no meaning so I excluded this.
