```python
# My visualization of choice is the barplot (https://python-graph-gallery.com/barplot/)

# Part I. 

# Libraries
import numpy as np
import matplotlib.pyplot as plt

# Make a random dataset:
height = [3, 12, 5, 18, 45]
bars = ('A', 'B', 'C', 'D', 'E')
y_pos = np.arange(len(bars))

# Create bars
plt.bar(y_pos, height)

# Create names on the x-axis
plt.xticks(y_pos, bars)

# Show graphic
plt.show()
```

Part II. 

Aesthetic - Yes, the visualization is pleasing to look at. It is simple and clearly dispalyed. There is no confusing shapes or clashing colours. 

Substantive - Yes, to a degree. The axis has a clear range and bars are clearly labelled. But, the x and y axis do not have a title and I am unsure in this example what data is being presented. The visualization does not fully accurately and honestly present data. There is also no legend or any indication of what information we are supposed to be comparing.  

Perceptual - Yes, we can understand the message the maker of the visualization is attempting to convey. We see that bar E has the highest number and A is the lowest. But, we do not know what data the graph pertains to so the message is not completely clear.
